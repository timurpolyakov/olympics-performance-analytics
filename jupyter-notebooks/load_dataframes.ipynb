# Load Prepared DataFrames
This notebook loads the processed DataFrames and ML-ready preprocessed feature DataFrames saved by the project scripts.
Files loaded: `data/ml/train_df.pkl`, `data/ml/test_df.pkl`, `data/ml/X_train_df.pkl`, `data/ml/X_test_df.pkl`, and `data/ml/preprocessor.joblib`.

```python
import os
import pandas as pd
from IPython.display import display

base = os.path.join('..', 'data', 'ml')
paths = {
    'train_df': 'train_df.pkl',
    'test_df': 'test_df.pkl',
    'X_train_df': 'X_train_df.pkl',
    'X_test_df': 'X_test_df.pkl',
}
dfs = {}
for name, fname in paths.items():
    p = os.path.join(base, fname)
    if os.path.exists(p):
        try:
            dfs[name] = pd.read_pickle(p)
        except Exception as e:
            print(f'Failed to load {p}: {e}')
    else:
        print(f'Missing file: {p}')

for name, df in dfs.items():
    print(f'--- {name} ---')
    print('shape:', df.shape)
    display(df.head())

# quick check: class balance in preprocessed sets if available
if 'X_train_df' in dfs:
    print('Train medal balance:', dfs['X_train_df']['medal_binary'].value_counts(normalize=True).rename('proportion'))
```

```python
# load preprocessor (pipeline) to inspect it
import joblib
preproc_path = os.path.join(base, 'preprocessor.joblib')
if os.path.exists(preproc_path):
    pipe = joblib.load(preproc_path)
    print(type(pipe))
    try:
        print('Pipeline steps:', pipe.named_steps)
    except Exception:
        pass
else:
    print('Preprocessor not found at', preproc_path)
```

Logistic Regression

```python
X_train_df = dfs.get('X_train_df')
X_test_df = dfs.get('X_test_df')

X_train_F = X_train_df[X_train_df['Sex_F'] == 1.0]
y_train_F = X_train_F['medal_binary']
X_train_F = X_train_F.drop(columns=['medal_binary'])

X_train_M = X_train_df[X_train_df['Sex_M'] == 1.0]
y_train_M = X_train_M['medal_binary']
X_train_M = X_train_M.drop(columns=['medal_binary'])


X_test_F = X_test_df[X_test_df['Sex_F'] == 1.0]
y_test_F = X_test_F['medal_binary']
X_test_F = X_test_F.drop(columns=['medal_binary'])

X_test_M = X_test_df[X_test_df['Sex_M'] == 1.0]
y_test_M = X_test_M['medal_binary']
X_test_M = X_test_M.drop(columns=['medal_binary'])


print('X_train_F shape:', X_train_F.shape)
print('y_train_F shape:', y_train_F.shape)
print('X_test_F shape:', X_test_F.shape)
print('y_test_F shape:', y_test_F.shape)


```

Model For Women's Sports

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import accuracy_score

model_F = LogisticRegression(class_weight='balanced', max_iter=1000)

model_F.fit(X_train_F, y_train_F)

y_pred_F = model_F.predict(X_test_F)
y_prob_F = model_F.predict_proba(X_test_F)[:, 1]

print(classification_report(y_test_F, y_pred_F))
print('Confusion Matrix:\n', confusion_matrix(y_test_F, y_pred_F))
print('Accuracy of Model: ',accuracy_score(y_test_F, y_pred_F))
print(classification_report(y_test_F, y_pred_F))

```

```python
coef_df_F = pd.DataFrame({
    'Feature': X_train_F.columns,
    'Coefficient': model_F.coef_[0]
})

coef_df_F.sort_values(by='Coefficient', ascending=False)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

model_F2 = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',
    random_state=42
)

model_F2.fit(X_train_F, y_train_F)

y_pred_F2 = model_F2.predict(X_test_F)
y_prob_F2 = model_F2.predict_proba(X_test_F)[:, 1]

print("Accuracy:", accuracy_score(y_test_F, y_pred_F2))
print(classification_report(y_test_F, y_pred_F2))
print(confusion_matrix(y_test_F, y_pred_F2))
```

```python
def recommend_sport(model, athlete_row, sport_cols):
    results = []

    for sport in sport_cols:
        athlete_copy = athlete_row.copy()

        athlete_copy[sport_cols] = 0
        athlete_copy[sport] = 1

        athlete_df = pd.DataFrame([athlete_copy])

        prob = model.predict_proba(athlete_df)[0, 1]

        results.append({
            "Sport": sport.replace("Sport_", ""),
            "Medal_Probability": prob
        })

    results_df = pd.DataFrame(results)

    best_row = results_df.loc[results_df["Medal_Probability"].idxmax()]

    return best_row, results_df.sort_values("Medal_Probability", ascending=False)
```
